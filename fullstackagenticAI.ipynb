```python
import gradio as gr
import base64
from io import BytesIO

def analyze_art(image, question):

    if image is None:
        return "Please upload a painting first."

    if not question.strip():
        return "Please enter a question for your art mentor."

    # Convert PIL image to base64
    buffered = BytesIO()
    image.save(buffered, format="PNG")
    image_base64 = base64.b64encode(buffered.getvalue()).decode("utf-8")

    # Send image + question to OpenAI
    response = client.responses.create(
        model="gpt-4.1-mini",
        input=[
            {
                "role": "user",
                "content": [
                    {
                        "type": "input_text",
                        "text": f"""
You are an experienced watercolor and visual-art mentor.

Analyse the uploaded painting and answer the user's question.

Focus on:
- composition
- colour harmony
- values and contrast
- focal point
- edges
- visual balance
- areas that can be improved

Give practical, constructive suggestions.

User question:
{question}
"""
                    },
                    {
                        "type": "input_image",
                        "image_url": f"data:image/png;base64,{image_base64}"
                    }
                ]
            }
        ]
    )

    return response.output_text


with gr.Blocks() as demo:

    gr.Markdown("# 🎨 AI Art Mentor Agent")

    image = gr.Image(
        type="pil",
        label="Upload your painting"
    )

    question = gr.Textbox(
        label="Ask your mentor",
        placeholder="e.g. Please analyse and suggest improvements to this painting."
    )

    submit = gr.Button("Analyze")

    output = gr.Markdown(
        label="AI Art Mentor Feedback"
    )

    submit.click(
        fn=analyze_art,
        inputs=[image, question],
        outputs=output
    )

demo.launch()
```
